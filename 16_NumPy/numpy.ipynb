4) Indexing, Slicing, and Views

Slicing usually returns a view, not a copy, A view shares memory with the original array. This is fast, but it means edits can affect the souce array.

```python
x = np.arange(12).reshape(3,4)
print("original array: \n", x)
```

```python
# Select one row: (1st row)
print("\n first row\n:" , x[0])
```

```python
# selecting 1 column (2nd column)
print("\n second column\n:", x[:, 2])
```

```python
# select a submatrix: rows 0 and 1, columns 1 and 2

sub = x[:2, 1:3]
print("submatrix \n", sub)
```

```python
# because sub is a view, this edir changes x too
sub[0,0] = 999
print("original x\n", x)
```

#### Deep Copy

#### since when we cahnge anything it will get changed in both original and the view to avoid that we use Deep copy

```python
# since when we cahnge anything it will get changed in both original and the view to avoid that we use Deep copy

# use .copy() when you need independent data

safe_sub = x[:2, 1:3].copy()
print("\n safe submatrix copy before edit: \n", safe_sub)
safe_sub[0,0] = -1
print("\n safe submatrix copy after edit:\n", safe_sub)
print("\n original array\n", x)
```

5) Boolean Masks and conditional Selection

- Boolean masks make it possible to filter and edit arrats without explicit loops. This is one of the most common array-processing patterns.

```python
scores = np.array([48,76,92,35,68,88,59])
print("original array\n", scores)
```

```python
# build the condition for values that meet a threshold.

passed = scores >= 60
print("masked", passed)
print("selected scores", scores[passed])
```

```python
scores = np.array([48,76,92,35,68,88,59])

# np.where chooses values element by element

labels = np.where(scores >= 60, "review-later", "needs-attention")
print("Labels: \n" , labels)
```

6) Vectorization: replacing Loops with Array expressions

- vectorization means expressing operations over whole arrays at once. It usually shorter, clearer, and faster than manual Python loops because NumPy does the heavy work in optimized complied code.

```python
values = rng.normal(loc=100, scale=15, size=10_00)
# loc - mean of the distribution, scale - standard deviation, size - number of samples
# Loop style: correct, but slower and more verbose for large arrays

centered_loop = []
mean_value = values.mean()
for value in values:
    centered_loop.append(value - mean_value)
centered_loop = np.array(centered_loop)

# vectorised style: subtract one scalar from every element
centered_vectorized = values - values.mean()

#allclose is safer than exact equality for floating-point values
print("same result:", np.allclose(centered_loop, centered_vectorized))
print("mean after centering", centered_vectorized.mean().round(10))
```

#### Broadcasting: Shape- Aware Arithmetic

- broadcating lets NumPy combine arrays with different, but compatible, shapes. This is essential for column-wise scaling, row-wise offsets, batch operations, and tensor-style calculations.

```python
x = np.array([
    [170,65,23],
    [182,82,31],
    [160,52,28],
    [175,74,35],
], dtype=float)

print("original data:\n", x)


column_means = x.mean(axis=0)
column_stds = x.std(axis=0)

x_scaled = (x - column_means) / column_stds

print("column mean: ", column_means)
print("column stds: ", column_stds)
print("\n scaled data: \n", x_scaled.round(3))
print("\n scaled column means: ", x_scaled.mean(axis=0).round(10))
print("\n scaled column stds: ", x_scaled.std(axis=0).round(10))
```

#### Reshaping and axis discipline

- Shape mistakes are among the most common causes of numerical bugs. Learn to reshape intentionally and use keepdims=True when a reduced result must still broadcast cleanly.


```python
a = np.arange(24)
print("original array: \n", a)

# reshape without changing data order
batch = a.reshape(2,3,4)

# sum over the last axis, keepdims=true preserves a singleton axis.
row_sums = batch.sum(axis=1, keepdims=True)
print("rows sums shape:", row_sums.shape)

# because row_sums has shape (2,3,1) it broadcasts across the last axis.
normalized_rows = batch / np.where(row_sums == 0, 1, row_sums)
print("\n first matrix row-normalized: \n", normalized_rows[0])

# add a new axis explicitly when needed.
column_vector = np.arange(3)[: , None]
row_vector = np.arange(4)[None, :]
print("\n outer-style sum: \n", column_vector + row_vector)
```

#### Handling Missing and Invalid Values

- Real arrays often contains missing or invalid values. NumPy represents missing floating-point values with np.nan. Use nan-aware functions when appropriate.

```python
readings =  np.array([
    [10.0, 12.0, np.nan],
    [11.0, np.nan, 15.0],
    [9.0, 13.0, 14.0]
])

print("raw reading \n", readings)

#standard mean returns nan if any value in the slice is nan
print("\n standard column mean: ", readings.mean(axis=0))

#nanmean ignores nan values.
col_means = np.nanmean(readings, axis=0)
print("nan aware column mean", col_means)

# replace missing values with the matching column mean
imputed = np.where(np.isnan(readings), col_means, readings)
print("\n inputed readings: \n", imputed)

# detect finite values. this catches nan, +inf, and -int
print("\n all values finit after imputation: ", np.isfinite(imputed).all())
```

### Sorting, ranking, Top-k Selection

- Full sorting is useful, but sometimes you only need the largest or smallest few values. argpartition can be more efficient for top-k style selection

```python
relevance = rng.uniform(0,1, size=12)
print("relevance scores: ", relevance.round(3))
k = 4

# argsort returns indices that would sort the array
sorted_indices = np.argsort(relevance)[::-1]
print("scores", relevance.round(3))
print("top indices by full sort: ", sorted_indices[:k])
```
