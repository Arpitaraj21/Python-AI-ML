### LIST VS NumPy array

```python
my_list = [1,2,3]

print("Python list multiplication", my_list * 2)


np_array = np.array([1,2,3]) #element wise multiplcation
print("Python np array multiplication", np_array * 2)


# to check the effiency who will get execute first
import time 
start = time.time()
my_list = [i * 2 for i in range (1000000)]
print("\n list operation time: " , time.time() - start)

start = time.time()
np_array = np.arange(1000000) * 2
print("\n np array operation time: " , time.time() - start)
```

### Array Reshaping

```python
array = np.arange(12)
print("original array : " , array)

reshaped_array = array.reshape((3,4)) # convert to 3 x 4 matric
print("reshaped array", reshaped_array)


flattened = reshaped_array.flatten()
print("flattend array", flattened) # returns copy of the array

# important
# returns view instead of copy (returns orignal array)
raveled = reshaped_array.ravel()
print("raveled", raveled)

# transpose

transpose = reshaped_array.T
print("transpose array", transpose)
```
