# Practice Questions

### Basic

```python
import numpy as np
```

```python
# create array 1 to 20
arr = np.arange(0,20)
arr
```

```python
# create 3 x 3 zero matrix

np.zeros((3, 3))
```

```python
# find shape, size, dtype

# shape
arr.shape

# size
arr.size

# dtype
arr.dtype
```

```python
# convert list - array

my_list = [1,2,3,4,5]
arr = np.array(my_list)
arr
```

### Indexing

```python
# Get even numbers only

even_arr = arr[arr % 2 == 0]
even_arr
```

```python
# replace negative with 0
arr[arr < 0] = 0
arr
```

```python
# extract first 5 elements

arr[0:5]
```

```python
# fancy indexing 

arr[[1,5]]
```

### Reshape

```python
# reshape (12,) → (3,4)

arr = np.arange(0,12)
arr
```

```python
array = arr.reshape(3,4)
```

```python
arr = np.array([[1,2],[3,4]])

flatten_array = arr.flatten()
flatten_array
```

```python
arr.ravel()
```

```python
arr.reshape(-1,1)
```

### Stats

```python
# mean, median, std, var

arr.mean()
np.median(arr)
np.std(arr)
np.var(arr)
```

```python
# percentiles (25, 50, 90)

np.percentile(arr, 25)
```

```python
# np.percentile(arr, 50)

np.percentile(arr, 50)
```

```python
# np.percentile(90)

np.percentile(arr, 90)
```
