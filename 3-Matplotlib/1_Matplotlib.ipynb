````python
import matplotlib.pyplot as plt

# pyplot is a collection of functions that make matplotlib work like MATLAB. Each pyplot function makes some change to a figure: e.g., creates a figure, creates a plotting area in a figure, plots some lines in a plotting area, decorates the plot with labels, etc.
````

### Create first chart

````python
x = [1,2,3,4,5]
y =[10,20,25,35,45]


# line chart
plt.plot(x,y)
plt.show() # to display in juypter notebook
````

### customized chart

````python
plt.figure(figsize=(6,4))  #figure size

plt.plot(x,y,color='green', marker='o', linestyle='dashed', linewidth=2, markersize=12)
plt.title("Title of chart")
plt.xlabel("will take x-axis label")
plt.ylabel("will take y label")
plt.show()
````

### Advanced Line Chart - multiple lines & Legends

````python
x = [1,2,3,4,5]
y1 =[10,20,25,35,45]
y2 = [20,30,35,45,55]

plt.plot(x,y1, label="Sales 2024")
plt.plot(x,y2, label="Sales 2025")

plt.title("YoY Sales")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.legend() #to show legends
plt.show()
````

### Bar Charts

````python
x = [1,2,3,4,5]
y =[10,20,25,35,45]

plt.bar(x,y)
plt.title("Bar Chart")
plt.show()
````

### Histogram

````python
#data
# used for distribution analysis

import random
data = [random.randint(1,10) for _ in range(100) ]

plt.hist(data, bins=5) #bins means bar
plt.title("Histogram Example")
plt.show()
````

### Pie Chart

##### used to show part-to-whole relationships/data

````python
categories = ['A', 'B' , 'C', 'D', 'E' ]
sales = [10, 20, 30, 40, 45]

plt.pie(sales, labels=categories, autopct='%1.1f%%', startangle=90)
plt.title('Pie Chart')
plt.show()
````

### Scatter Plot

##### mostly used to find relationship between variables, 

````python
y1 =[10,20,25,35,45]
y2 = [20,30,35,45,55]

plt.scatter(y1, y2)
plt.title("Scatter Plot")
plt.show()
````

### Subplots

#### used to show multiple charts in one figure

````python
#data-1 (bar chart)
categories = ['A', 'B' , 'C', 'D', 'E' ]
sales = [10, 20, 30, 40, 45]

#data-2 (scatter chart)
y1 =[10,20,25,35,45]
y2 = [20,30,35,45,55]


plt.figure(figsize=(10,6))

#first plot  bar chart
plt.subplot(1,2,1) #1 row, 2 columns, 3rd position
plt.bar(categories, sales)
plt.title("Daily sales")
plt.xlabel("Week Days")
plt.ylabel("Sales")

# second plot - scatter chart

plt.subplot(1,2,2)
plt.scatter(y1,y2)
plt.title("User Example")
plt.xlabel("User1")
plt.ylabel('User2')

plt.show()
````

````python
import pandas as pd

#data

data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr'],
    'Sales': [12000, 11000, 13000, 25000]
}

df = pd.DataFrame(data)
df
````

### Obejct Oriented Interface

##### In Matplotlib, the Object-Oriented (OO) interface uses Figure and Axes objects instead of directly calling plt.bar(), plt.plot(), etc. 

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 25, 35]

fig, ax = plt.subplots()
ax.bar(x,y)
ax.set_title("Bar Chart")
ax.set_xlabel("X values")
ax.set_ylabel("Y Values")
````

## OO Interface VS Pyplot Interface

| Pyplot          | Object-Oriented   |
| --------------- | ----------------- |
| `plt.plot()`    | `ax.plot()`       |
| `plt.bar()`     | `ax.bar()`        |
| `plt.scatter()` | `ax.scatter()`    |
| `plt.title()`   | `ax.set_title()`  |
| `plt.xlabel()`  | `ax.set_xlabel()` |
| `plt.ylabel()`  | `ax.set_ylabel()` |
