### Aggregation and group by

```python
df['Team'].value_counts()
```

```python
df[df['Team'] == 'HR'].value_counts()
```

```python
#  Groupby

df.groupby('Team')['Monthly_Salary'].sum() #aggregation based on group by
```

```python

#  for avg we need to use mean

df.groupby('Team')['Monthly_Salary'].mean()
```

```python
df.groupby('Team').agg({'Monthly_Salary': 'mean', 'Name': 'count'})
```

### Concatenate and Merge Dataframe (JOINS)

```python
df1 = pd.DataFrame({'ID': [1,2,3], 'Name': ['A', 'B', 'C']})
df1
```

```python
df2 = pd.DataFrame({'ID': [1,2,3,4], 'Score': [88,96,77,79]})
df2
```

```python
pd.concat([df1, df2], axis=0) #vertical / row level join - top on top
```

```python
pd.concat([df1, df2], axis=1) # horizontal / column level join - left to right
```

```python
# merge = join

pd.merge()
```
