## Indexing, Selecting & Assigning

```python
reviews = pd.DataFrame({
    'country': ['Italy', 'France', 'Portugal'],
    'description': ['Excellent wine', 'Great taste', 'Delicious'],
    'points': [95, 87, 92],
    'price': [15.99, 12.99, 18.99],
    'province': ['Tuscany', 'Bordeaux', 'Douro'],
    'region_2': ['Chianti', 'Medoc', 'Douro']
})

reviews
```

```python
reviews.country
```

If we gave a python dictionary , we can access its values using the indexing ([]) operator. We can do the same with columns in a DataFrame.

```python
reviews['country']
```

```python
reviews['country'][0]
```

iloc - index based selection

```python
reviews.iloc[0]
```

```python
reviews.iloc[:, 0]
```

```python
reviews.iloc[1:3, 0]
```

### it is possible to pass a list

```python
reviews.iloc[[0,1,2], 0]
```

 negative numbers - start counting forwards from the end of the values

```python
reviews.iloc[-5:]
```

label-based selection

```python
reviews.loc[0, 'country']
```

```python
reviews.loc[:, ['taster_name', 'taster_twitter_handle', 'points']]
```
