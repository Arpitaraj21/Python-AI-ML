 Without dict comprehensive
 code to find the squares or numbers from 1 to 10 and store them in a dictionary without using dictionary comprehension

```python
squares = {}

for i in range(1,11):
    squares[i] = i ** 2
print(squares) 

print({i: i ** 2 for i in range(1,11)})
```

```python
# dict compreshension with if  condition:
# code to find the squares of even numbers from 1 to 10 and store them in a dict using dict comprehension with an if condition.

squares = {}
for i in range(1,11):
    if i % 2 == 0:
        squares[i] = i ** 2

print(squares)


print({i: i ** 2 for i in range(1,11) if i % 2 == 0})
```

```python
# dict compreshension with if-else  condition:
# code to find the squares of even numbers from 1 to 10 and store them in a dict using dict comprehension with an if condition.

squares = {}
for i in range(1,11):
    if i % 2 == 0:
        squares[i] = i ** 2
    else:
        squares[i] = i ** 3

print(squares)


# dict comprehensive :
# syntax:
# (key_expression: value_expression if condition else value_expression for item in iterable)
print({i: i ** 2 if i % 2 == 0 else i ** 3 for i in range(1,11)})
```

# Lambda Function

Q) What is a Lambda function?
-> A lambda function is a small, anonymous function ( no def, no name).

- written in one line
- used when we need a simple, short function without formally defining it
- common in data processing, sorting, filtering, mapping and machine learning pipelines

```python
# User-defined function to find the square of a number

def sqr(no):  # no. - argumnent
    return no * no  # no * no - expression

z = sqr(5)
print(z)
```

- lambda function
- syntax:
- lambda arguments: expression
- arguments are inputs (like in a normal function)
- expression is a single output line (no multiple statements allowed)

```python
z = lambda no: no * no

print(z(5))
```

```python
# example2: adding two numbers using lambda function
# lambda arguments: expression

add = lambda a,b: a + b
print(add(2,3))
```

```python
#Example3: finding maximum of two numbers using lambda functions
# lambda arguments: expression
# maxn = lambda x,y : x if x > y else y

maxn = lambda x,y: x if x > y else y
# maxn = lambda x,y: max(x,y)
maxium = maxn(15,20)
print(f"The number is {maxium}")
```

# Map, Filter , Reduce

- Map, filter and reduce are built-in functions of python
- These functions enable the functional programming aspect of python.
- In functional programming, the arguments passed are the only factors that decide upon the output
- These functions can take any other function as a parameter and can supplied to other functions as parameters as well.

map()
- It applies the given function to all iterables and return a new list.

syntax: map(function, iterables)

```python
# without map
def sqr(a):
    return a * a

x = sqr(5)
print(x)
```

```python
#with map
# map(function, iterables)

def sqr(a):
    return a * a

x = map(sqr, [1,2,3,4,5])

print(x)
print(list(x))
```

```python
def sqr(n):
    return n * n

x = map(sqr, [1,2,3,4,5])

print(x)
print(tuple(x))
# ans - tuple format will come
```

```python
def multipi(x,y):
    return x * y

z = map(multipi, [1,2,3], [4,5,6])
print(z)
print(list(z))
```
