```python
names = ["Arpita", "Darshan", "Neeraj", "Shubham", "Siddharth"]
phy_marks = [90,80,70,60,50]
chem_marks = [80,70,60,50,40]
maths_marks = [70,60,50,40,87]

# Access Arpita's marks
print(names[0], phy_marks[0], chem_marks[0], maths_marks[0])

#Access Darshan's marks
print(names[1], phy_marks[1], chem_marks[1], maths_marks[1])
```

```python
# create class
class Student:
    name = ""
    phy_marks = 0
    chem_marks = 0
    maths_marks = 0

    def total_marks(self):
        return self.phy_marks + self.chem_marks + self.maths_marks


# create obejct
p1 = Student()
p1.name = "Arpita"
p1.phy_marks = 90
p1.chem_marks = 80
p1.maths_marks = 70


p2 = Student()
p2.name = "Darshan" 
p2.phy_marks = 80
p2.chem_marks = 70
p2.maths_marks = 60


# Access Arpita's marks
print(f"Name: {p1.name}, Physics: {p1.phy_marks}, Chemistry: {p1.chem_marks}, Maths: {p1.maths_marks}, total marks: {p1.total_marks()}")


# total marks of Arpita
# total_marks_p1 = p1.phy_marks + p1.chem_marks + p1.maths_marks
print(f"Total marks of {p1.name} is {p1.total_marks()}")

# total marks of Darshan
# total_marks_p2 = p2.phy_marks + p2.chem_marks + p2.maths_marks
print(f"Total marks of {p2.name} is {p2.total_marks()}")
```

### Create a class make object and a function to find percentage

```python
# create a class

class Students:
    name = ""
    phy_marks = 0
    chem_marks = 0
    maths_marks = 0

    def marks_percentage(self):
        return round(((self.phy_marks + self.chem_marks + self.maths_marks) / 300)  * 100, 2)
    

# create one object

p1 = Students()

p1.name = "Arpita Raj"
p1.phy_marks = 90
p1.chem_marks = 60
p1.maths_marks = 70

print(f"Percentage of {p1.name} is {p1.marks_percentage()}%")
```

### Constructor
- A constructor is a special method that is called automatically when an object is created
- It is used to initialize the attributes of the object
- It is defined using the __init__() method
- It is called automatically when an object is created
- It takes self as the first parameter
- It can take any number of parameters

```python
class ABC:
    def __init__(self):
        self
        print("This is constructor")

    def greet(self):
        print("hello world!")


o1 = ABC()
o1.greet()
```

```python
# constructor with parameters
class ABC:
    def __init__(self, name):
        self.name = name
        print("constructor called")

    def greet(self):
        print(f"hello {self.name}!")
    

o1 = ABC("Arpita")
o1.greet()
```
